```python
import os
import re
import pandas as pd
import numpy as np
import emoji
import string
import nltk
from attacut import tokenize
from pythainlp import word_tokenize
from pythainlp.corpus.common import thai_stopwords
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.preprocessing import normalize
from sqlalchemy import create_engine, text
from dotenv import load_dotenv
from gensim.models import FastText
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords
```

```python
FINE_TUNED_DIR = '../pretained_or_finetune-models'
NLTK_DATA_PATH = f"{FINE_TUNED_DIR}/nltk_data"

nltk.data.path.append(NLTK_DATA_PATH)

stop_words = set(stopwords.words('english')) 
lemmatizer = WordNetLemmatizer()

BASE_DIR = "."
DATASET_DIR = os.path.join(BASE_DIR, "inputs")


def load_data_from_gcs(filename: str):
    # url = f"gs://my-streamlit-data/inputs/{filename}"
    url = f"{DATASET_DIR}/{filename}"
    return pd.read_parquet(url, engine="pyarrow")
```

```python
tripadvisor_reviews_sentiment = load_data_from_gcs("sentiment_prediction.parquet")
tripadvisor_attractions_details = load_data_from_gcs("combined_details.parquet")
attractions_tags_cluster = load_data_from_gcs("cosine_clusters_V3.parquet")
tat_attractions = load_data_from_gcs("final_tat_attractions.parquet")
maps_tat_and_tripadvisor_ids = load_data_from_gcs("TripAdvisor_Location_Ids.parquet")

if tripadvisor_reviews_sentiment.empty or attractions_tags_cluster.empty or tat_attractions.empty:
    raise ValueError("❌ One or more required tables are empty. Please check your database.")
```

```python

def clean_english_text(text):
    text = text.lower()
    text = re.sub(r'\d+', '', text)  # Remove numbers
    text = re.sub(r'[^a-z\s]', '', text)  # Remove special characters
    text = re.sub(r'\b(u|ur|b4)\b', 'you', text)  # Replace common abbreviations
    
    # Lowercasing
    text = text.lower()
    
    # Handling Emoji
    text = emoji.demojize(text)
    
    # Removing Punctuation
    text = ''.join([char for char in text if char not in string.punctuation])
    
    # Tokenization
    tokens = nltk.word_tokenize(text)
    
    # Removing Stop Words
    stop_words = set(stopwords.words('english'))
    tokens = [word for word in tokens if word not in stop_words]

    # Lemmatization
    lemmatizer = WordNetLemmatizer()
    tokens = [lemmatizer.lemmatize(token) for token in tokens]

    # Join tokens back to a single string
    return ' '.join(tokens)

def clean_thai_text(text):
    """Preprocess Thai text by removing non-Thai characters and stopwords."""
    if not isinstance(text, str):
        return ""
    stopwords_list = set(thai_stopwords())
    text = re.sub(r"[^ก-๙\s]", "", text)  # Remove non-Thai characters
    tokens = word_tokenize(text)
    cleaned_tokens = [word for word in tokens if word not in stopwords_list]
    return " ".join(cleaned_tokens)

def vectorize_fasttext(text, model, vector_size=None):
    if not isinstance(text, str) or not text.strip():
        return np.zeros(model.vector_size if vector_size is None else vector_size, dtype=np.float32)

    try:
        tokens = tokenize(text)  # สมมุติว่า tokenize ดีแล้ว
    except Exception as e:
        print(f"⚠️ Tokenization error: {e}")
        return np.zeros(model.vector_size if vector_size is None else vector_size, dtype=np.float32)

    vectors = [model.wv[word] for word in tokens if word in model.wv.key_to_index]

    if not vectors:
        return np.zeros(model.vector_size if vector_size is None else vector_size, dtype=np.float32)

    return np.mean(vectors, axis=0)
```

```python
filtered_attractions_df["cleaned_introduction_th"] = filtered_attractions_df["introduction_th"].apply(lambda x: clean_thai_text(x) if isinstance(x, str) else "")
filtered_attractions_df["cleaned_place_name_th"] = filtered_attractions_df["place_name_th"].apply(lambda x: clean_thai_text(x) if isinstance(x, str) else "")

filtered_attractions_df["merged_attraction_content_th"] = (
    filtered_attractions_df["cleaned_introduction_th"].fillna("") + " " +
    filtered_attractions_df["cleaned_place_name_th"].fillna("")
).str.strip()

filtered_attractions_df["cleaned_introduction_en"] = filtered_attractions_df["introduction_en"].apply(lambda x: clean_english_text(x) if isinstance(x, str) else "")
filtered_attractions_df["cleaned_place_name_en"] = filtered_attractions_df["place_name_en"].apply(lambda x: clean_english_text(x) if isinstance(x, str) else "")

filtered_attractions_df["merged_attraction_content_en"] = (
    filtered_attractions_df["cleaned_introduction_en"].fillna("") + " " +
    filtered_attractions_df["cleaned_place_name_en"].fillna("")
).str.strip()
```

```python
TH_filtered_attractions_df = filtered_attractions_df[filtered_attractions_df["merged_attraction_content_th"].str.strip() != ""]
EN_filtered_attractions_df = filtered_attractions_df[filtered_attractions_df["merged_attraction_content_en"].str.strip() != ""]

TH_tokenized_sentences = [word_tokenize(text) for text in TH_filtered_attractions_df["merged_attraction_content_th"] if isinstance(text, str) and text.strip()]
EN_tokenized_sentences = [word_tokenize(text) for text in EN_filtered_attractions_df["merged_attraction_content_en"] if isinstance(text, str) and text.strip()]
```

```python
TH_attraction_vectors = []

if not TH_tokenized_sentences:
    print("❌ No tokenized sentences available for Fasttext training. Ensure preprocessing is correct.")
else:
    TH_fasttext_model = FastText(
        vector_size=300,
        window=10,
        min_count=2,
        workers=4,
        sg=1,
        min_n=2,
        max_n=6,
        negative=10
    )
    if TH_tokenized_sentences:
        TH_fasttext_model.build_vocab(TH_tokenized_sentences)
        TH_fasttext_model.train(TH_tokenized_sentences, total_examples=TH_fasttext_model.corpus_count, epochs=20)
    else:
        print("⚠️ No TH data for FastText training.")
        
TH_attraction_vectors = np.array([
    vectorize_fasttext(text, TH_fasttext_model) 
    for text in filtered_attractions_df["merged_attraction_content_th"]
], dtype=np.float32)
```

```python
EN_attraction_vectors = []

if not EN_tokenized_sentences:
    print("❌ No tokenized sentences available for Fasttext training. Ensure preprocessing is correct.")
else:
    EN_fasttext_model = FastText(
        vector_size=300,
        window=5,  # อังกฤษใช้ window เล็กกว่าหน่อย
        min_count=2,
        workers=4,
        sg=1,
        min_n=2,
        max_n=6,
        negative=10
    )

    if EN_tokenized_sentences:
        EN_fasttext_model.build_vocab(EN_tokenized_sentences)
        EN_fasttext_model.train(EN_tokenized_sentences, total_examples=EN_fasttext_model.corpus_count, epochs=20)
    else:
        print("⚠️ No EN data for FastText training.")
        
EN_attraction_vectors = np.array([
    vectorize_fasttext(text, EN_fasttext_model) 
    for text in filtered_attractions_df["merged_attraction_content_en"]
], dtype=np.float32)
```

```python
input_text ="ร่มรื่นธรรมชาติ"
language="TH"

cleaned_input_text = ""
if (language == "TH" and TH_attraction_vectors.size == 0) or (language == "EN" and EN_attraction_vectors.size == 0):
    print("❌ No vectors available for similarity search!")


if language == "TH":
    attraction_vectors = TH_attraction_vectors
    fasttext_model = TH_fasttext_model
    cleaned_input_text = clean_thai_text(input_text)
elif language == "EN":
    attraction_vectors = EN_attraction_vectors
    fasttext_model = EN_fasttext_model
    cleaned_input_text = clean_english_text(input_text)
else:
    raise ValueError("❌ Unsupported language. Use 'TH' or 'EN'.")
```

```python
input_vector = vectorize_fasttext(cleaned_input_text, fasttext_model)
input_vector = normalize(input_vector.reshape(1, -1))  # Normalize input vector

similarity_scores = cosine_similarity(input_vector, attraction_vectors)[0]
```

```python
top_5_indices = np.argsort(similarity_scores)[::-1][:5]

print("🔍 Top 5 Indices:", top_5_indices)
print("✅ Similarity Scores:", similarity_scores[top_5_indices])
```

```python
results = []
for idx in top_5_indices:
    row = filtered_attractions_df.iloc[idx]

    place_id = row["place_id"]
    place_name = row["place_name_th"]
    similarity_score = similarity_scores[idx]
    merged_content = ""
    
    if language == "TH":
        merged_content = row["merged_attraction_content_th"]
    elif language == "EN":
        merged_content = row["merged_attraction_content_en"]
    else:
        merged_content = "N/A"
        
    result_review = maps_tat_and_tripadvisor_ids[maps_tat_and_tripadvisor_ids["place_id"] == place_id]
    location_id = result_review["location_id"].tolist() if not result_review.empty else []
    
    print(f"🔍 [Score: {similarity_score:.3f}] Place ID: {place_id} / Place Name {place_name} >>>>>>>>> [location_id: {location_id}]")
```
