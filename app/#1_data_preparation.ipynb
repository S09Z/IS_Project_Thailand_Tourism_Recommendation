```python
import hook.thai_tourism_data as tat
tourism_location = tat.ThaiTourismAPI("EN")

def getTourismLocationsByProvince(province_name, province_location, category):
    search_params = {
            "keyword": "",  # Example keyword for Kanchanaburi
            "location": province_location,
            "categorycodes": category,
            "province_name": province_name,
            "radius": 1000,
            "number_of_result": 100,
            "page_number": 1,
            "destination": province_name,
            "filter_by_update_date": "2019/09/01-2024/05/31"
        }
    tourism_data = tourism_location.fetch_search_location(**search_params)

    df = pd.DataFrame([vars(place) for place in tourism_data])
    return df
```

```python
json_files = glob.glob("./../dataset/reviews/*_TH_*.json")

all_reviews = []

for file in json_files:
    with open(file, "r", encoding="utf-8") as f:
        data = json.load(f)  # Load JSON
        
        # Step 2: Flatten nested JSON structure
        for location_reviews in data:  
            for location_id, reviews in location_reviews.items():
                for review in reviews:
                    flattened_review = {
                        "location_id": review.get("location_id"),
                        "review_id": review.get("id"),
                        "lang": review.get("lang"),
                        "published_date": review.get("published_date"),
                        "rating": review.get("rating"),
                        "helpful_votes": review.get("helpful_votes"),
                        "url": review.get("url"),
                        "review_text": review.get("text"),
                        "review_title": review.get("title"),
                        "trip_type": review.get("trip_type"),
                        "travel_date": review.get("travel_date"),
                        "username": review.get("user", {}).get("username"),
                        "user_location": review.get("user", {}).get("user_location", {}).get("name"),
                        "avatar_url": review.get("user", {}).get("avatar", {}).get("thumbnail"),
                    }
                    all_reviews.append(flattened_review)

# Step 3: Convert to Pandas DataFrame
merged_reviews = pd.DataFrame(all_reviews)
merged_reviews

# Save as Parquet with compression (choose one)
merged_reviews.to_parquet("./../dataset/reviews/TripAdvisor_reviews_TH.parquet", engine="pyarrow", compression="zstd") 
```

```python
tripadvisor_reviews = []

for k, location in tripadvisor_central_attraction_locations.iterrows():
    response = TripAdvisor_api.location_reviews(
        location['location_id'],
        language
    )
    
    try:
        reviews = response.json()['data']
        if len(reviews) == 0:
            print(f" #{k + 1} >>> [{language}] {location['location_id']}: No Data")
            continue
        
        review = {f"{location['location_id']}": reviews}
        tripadvisor_reviews.append(review)
        
        print(f" #{k + 1} >>> [{language}] {location['location_id']}")
        
    except KeyError as e:
        print(f" #{k + 1} >>> [{language}] {location['location_id']} ::: {e} {response}")
        continue
    
    # if (k == 5):
    #     break
```

```python
# Import necessary libraries
import geopandas as gpd
import matplotlib.pyplot as plt

# Load the GeoJSON file
geojson_file = 'hook/data/thai_province_coordinates_geo.json'
gdf = gpd.read_file(geojson_file)

# Check for empty geometries and remove them
gdf = gdf[~gdf.is_empty]

# Ensure the CRS is set to WGS 84 (EPSG:4326)
if gdf.crs is None:
    gdf.set_crs(epsg=4326, inplace=True)

# Initialize a Matplotlib figure and axis
fig, ax = plt.subplots(figsize=(12, 8))

# Plot all provinces in light blue
gdf.plot(ax=ax, color='lightblue', edgecolor='black')

# Highlight Bangkok in red, checking if it exists in the data
bangkok = gdf[gdf['name'] == 'Bangkok Metropolis']
if not bangkok.empty:
    bangkok.plot(ax=ax, color='red', edgecolor='black')

# Set the aspect of the plot to equal to ensure accurate representation
ax.set_aspect('equal')

# Add title and labels
ax.set_title('Map of Thai Provinces with Bangkok Highlighted', fontsize=15)
ax.set_xlabel('Longitude', fontsize=12)
ax.set_ylabel('Latitude', fontsize=12)

# Show the plot
plt.show()
```
